Dropping the unwanted Columns and Row's Values

```python
df.drop(columns=['S.No.','New_Price'],inplace=True)
df.dropna(subset=['Price'],inplace=True)
df.head(3)
```

Splitting the unwanted parts from Mileage,Engine and Power and converting it to float dtype

```python
df['Mileage'] = pd.to_numeric(df['Mileage'].str.split().str[0], errors='coerce')
df['Engine'] = pd.to_numeric(df['Engine'].str.split().str[0], errors='coerce')
df['Power'] = pd.to_numeric(df['Power'].str.split().str[0], errors='coerce')

df.head(3)
```

```python
df.isnull().sum()
```

```python
df['Mileage'] = df['Mileage'].fillna(df['Mileage'].median())
```

```python
df['Engine'] = df['Engine'].fillna(df.groupby('Name')['Engine'].transform('median'))
df['Power']  = df['Power'].fillna(df.groupby('Name')['Power'].transform('median'))
df['Engine'] = df['Engine'].fillna(df['Engine'].median())
df['Power']  = df['Power'].fillna(df['Power'].median())
```

```python
df['Seats'] = df['Seats'].fillna(df['Seats'].mode()[0])
```

```python
df.to_csv('used_cars_cleaned.csv', index=False)
```
